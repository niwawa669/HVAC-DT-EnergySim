```python
x = torch.tensor([1.0],  requires_grad=True)
```

```python
y = 2 * x
y.requires_grad
# y.retain_grad()
```

```python
z = y
print(y.requires_grad, z.requires_grad)
z.retain_grad()
z.add_(1)
z.retain_grad()
print(z.requires_grad, z.grad)
# z, y
```

```python
w = z**2
w
```

```python
w.backward()
```

```python
x.grad, y.grad, z.grad
```
